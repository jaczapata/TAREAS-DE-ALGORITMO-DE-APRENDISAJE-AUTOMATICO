## Selección de variables

Para el modelado se utilizarán todas las variables predictoras disponibles, excluyendo únicamente `Exam_Score`, debido a que esta es la variable objetivo.

La decisión permite aprovechar tanto las variables numéricas como las categóricas relacionadas con aspectos académicos, personales, familiares y escolares.

Durante la exploración se identificaron relaciones importantes entre `Exam_Score` y variables como `Attendance`, `Hours_Studied`, `Previous_Scores` y `Tutoring_Sessions`. Además, las variables categóricas pueden aportar información complementaria que no se observa mediante una correlación numérica.

Por esta razón, se conservarán los predictores disponibles y se utilizará un proceso de preprocesamiento reproducible.

```python
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE
)

print("Datos de entrenamiento:", X_train.shape)
print("Datos de prueba:", X_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Pipeline para variables numéricas
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Pipeline para variables categóricas
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

# ColumnTransformer
preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocesamiento creado correctamente.")
```

```python
from sklearn.dummy import DummyRegressor

baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", DummyRegressor(strategy="mean"))
])

baseline_model.fit(X_train, y_train)

baseline_predictions = baseline_model.predict(X_test)

print("Línea base entrenada correctamente.")
```

```python
from sklearn.linear_model import LinearRegression

linear_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)

print("Regresión lineal entrenada correctamente.")
```

```python
from sklearn.linear_model import Ridge

ridge_model = Pipeline([
    ("preprocessor", preprocessor),
    ("model", Ridge(alpha=10.0))
])

ridge_model.fit(X_train, y_train)

ridge_predictions = ridge_model.predict(X_test)

print("Modelo Ridge entrenado correctamente.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

def calculate_metrics(y_true, predictions):
    mae = mean_absolute_error(y_true, predictions)
    rmse = np.sqrt(mean_squared_error(y_true, predictions))
    r2 = r2_score(y_true, predictions)

    return mae, rmse, r2


baseline_metrics = calculate_metrics(
    y_test,
    baseline_predictions
)

linear_metrics = calculate_metrics(
    y_test,
    linear_predictions
)

ridge_metrics = calculate_metrics(
    y_test,
    ridge_predictions
)

results = pd.DataFrame({
    "Modelo": [
        "Línea base",
        "Regresión lineal",
        "Ridge"
    ],
    "MAE": [
        baseline_metrics[0],
        linear_metrics[0],
        ridge_metrics[0]
    ],
    "RMSE": [
        baseline_metrics[1],
        linear_metrics[1],
        ridge_metrics[1]
    ],
    "R²": [
        baseline_metrics[2],
        linear_metrics[2],
        ridge_metrics[2]
    ]
})

display(results.round(4))
```

## Interpretación de las métricas

El **MAE (Mean Absolute Error)** representa el error absoluto promedio de las predicciones. En este contexto, indica aproximadamente cuántos puntos de `Exam_Score` se aleja el modelo del puntaje real de un estudiante, en promedio.

El **RMSE (Root Mean Squared Error)** también se expresa en puntos de examen, pero da mayor peso a los errores grandes debido a que primero eleva los errores al cuadrado. Por esta razón, es útil para identificar si existen algunas predicciones con errores considerablemente altos.

El **R² (coeficiente de determinación)** representa qué proporción de la variabilidad de `Exam_Score` puede explicar el modelo. Un valor más cercano a 1 representa una mayor capacidad explicativa, mientras que un valor cercano a 0 indica que el modelo aporta poca mejora respecto a utilizar únicamente la media.

En este análisis, la regresión lineal y Ridge presentan una mejora considerable respecto a la línea base. Esto indica que las características disponibles sí contienen información útil para estimar el desempeño en el examen.

## Análisis de las observaciones con mayor y menor error

Las observaciones con mayor error absoluto representan estudiantes cuyo puntaje real resulta más difícil de estimar para el modelo. Particularmente, algunos casos presentan puntajes elevados que son subestimados.

Esto puede indicar que existen combinaciones particulares de características que el modelo no logra representar completamente. También puede deberse a variables relevantes que no están incluidas en el dataset o a relaciones no lineales entre los factores y el desempeño.

Por otro lado, las observaciones con menor error corresponden a casos que presentan características que el modelo puede representar con mayor facilidad.

Por lo tanto, los errores observados pueden estar relacionados tanto con las variables disponibles como con la estructura del modelo utilizado.

```python
from sklearn.inspection import permutation_importance

permutation = permutation_importance(
    ridge_model,
    X_test,
    y_test,
    n_repeats=10,
    random_state=RANDOM_STATE,
    scoring="r2"
)

importance = pd.Series(
    permutation.importances_mean,
    index=X_test.columns
).sort_values(ascending=False)

print("10 variables más importantes:")

display(
    importance
    .head(10)
    .to_frame("Importancia")
    .round(4)
)
```

```python
plt.figure(figsize=(8,5))

importance.head(10).sort_values().plot(
    kind="barh"
)

plt.title("Variables más relevantes para Ridge")
plt.xlabel("Disminución promedio de R²")

plt.tight_layout()
plt.show()
```

```python
from sklearn.model_selection import KFold, cross_val_score

kf = KFold(
    n_splits=5,
    shuffle=True,
    random_state=RANDOM_STATE
)

cv_scores = cross_val_score(
    ridge_model,
    X_train,
    y_train,
    cv=kf,
    scoring="neg_root_mean_squared_error"
)

cv_rmse = -cv_scores

print("RMSE de cada partición:")
print(np.round(cv_rmse, 4))

print("\nRMSE promedio:")
print(round(cv_rmse.mean(), 4))

print("\nDesviación estándar:")
print(round(cv_rmse.std(), 4))

print("\nRMSE de la prueba única:")
print(round(ridge_metrics[1], 4))
```

## Interpretación de la validación cruzada

La validación cruzada K-Fold divide el conjunto de entrenamiento en cinco particiones. En cada iteración, cuatro particiones se utilizan para entrenar el modelo y una para evaluarlo.

Este procedimiento permite observar cómo cambia el desempeño del modelo dependiendo de los datos utilizados para entrenamiento y evaluación.

La validación cruzada proporciona una referencia más estable que una sola división de entrenamiento y prueba, ya que el modelo se evalúa varias veces con diferentes particiones.

La media del RMSE representa el error esperado promedio durante las diferentes particiones, mientras que la desviación estándar indica cuánto varía el desempeño entre ellas.

La comparación con el RMSE obtenido en la partición de prueba permite identificar si la división utilizada inicialmente fue representativa o si pudo haber favorecido/desfavorecido al modelo.

# Conclusión

Después de entrenar y evaluar una línea base, una regresión lineal y un modelo Ridge, se encontró que **Ridge obtuvo el mejor desempeño**, aunque la diferencia con la regresión lineal fue pequeña.

En la partición de prueba, Ridge obtuvo aproximadamente un **MAE de 0.45**, lo que significa que sus predicciones se alejan en promedio alrededor de 0.45 puntos del puntaje real del examen. Su **RMSE fue aproximadamente 1.80**, mientras que su **R² fue cercano a 0.77**.

En comparación con la línea base, que simplemente predice la media de `Exam_Score`, los modelos de regresión presentan una mejora considerable. Esto demuestra que las características disponibles en el dataset contienen información útil para estimar el desempeño de los estudiantes.

Entre las variables que mostraron mayor relevancia se encuentran principalmente **Attendance** y **Hours_Studied**, además de variables como `Previous_Scores`, `Access_to_Resources`, `Parental_Involvement` y `Tutoring_Sessions`.

El análisis de residuos permitió identificar algunos casos en los que el modelo presenta errores mayores, principalmente al estimar algunos puntajes elevados. Esto puede deberse a factores no incluidos en el dataset o a relaciones no lineales que el modelo Ridge no logra representar completamente.

La validación cruzada K-Fold permitió comprobar que el desempeño puede variar dependiendo de la partición utilizada, por lo que proporciona una evaluación más estable que depender únicamente de una división de entrenamiento y prueba.

Como mejora futura sería conveniente probar modelos capaces de capturar relaciones no lineales, como **Random Forest, Gradient Boosting o regresión polinómica**, además de realizar una búsqueda de hiperparámetros.

En conclusión, el modelo Ridge puede considerarse una **primera aproximación útil para estimar `Exam_Score`**, pero no debería utilizarse por sí solo para tomar decisiones académicas importantes. Sería necesario realizar una validación adicional, analizar posibles sesgos e incorporar otras variables antes de utilizarlo como herramienta de apoyo para decisiones reales.
